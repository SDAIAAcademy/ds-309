```python
import importlib

missing = []
for name in ["pandas", "numpy", "sklearn", "matplotlib", "opendp", "fairlearn"]:
    try:
        mod = importlib.import_module(name)
        print(f"{name:12} {getattr(mod, '__version__', '?')}")
    except ImportError:
        missing.append(name)
print("Python", sys.version.split()[0])
```
